### Actividad 1: 
Suma de todos los elementos de una lista, utilizando bucles repetitivos (for, while).

```python
numbers = [14, 23, 34, 48, 85]

suma = 0

for num in numbers:
    suma += num

print(f"La suma de los elementos es: {suma}")

```

```python
numbers = [14, 23, 34, 48, 85]

suma = 0
i = 0

while i < len(numbers):
    suma += numbers[i]
    i += 1
print(f"La suma de los elementos es: {suma}")
```

### Actividad 2: 
Encontrar el número más grande en una lista, utilizando bucles repetitivos (for, while).

```python
from random import randint

random_numbers = [randint(1, 100) for i in range(5)]

print(f"Lista de números aleatorios: {random_numbers}")
```

```python
max_num = random_numbers[0]

for num in random_numbers:
    if num > max_num:
        max_num = num
        
print(f"El número más grande en la lista es: {max_num}")
```

```python
max_num = random_numbers[0]

i = 1

while i < len(random_numbers):
    if random_numbers[i] > max_num:
        max_num = random_numbers[i]
    i += 1
    
print(f"El número más grande en la lista es: {max_num}")
```

### Actividad 3:
Contar cuántas veces aparece un número específico. Utiliza estructuras repetitivas para la resolucion de este ejercicio. 

```python
from random import randint

lists = [randint(1, 200) for i in range(20)]

print(f"Lista de números: {lists}")
```

```python
numero_a_buscar = 50

contador = 0

for num in lists:
    if num == numero_a_buscar:
        contador += 1

print(f"El número {numero_a_buscar} aparece {contador} veces en la lista.")
```

```python
numero_a_buscar = 50 

contador = 0
i = 0

while i < len(lists):
    if lists[i] == numero_a_buscar:
        contador += 1
    i += 1

print(f"El número {numero_a_buscar} aparece {contador} veces en la lista.")
```

### Actividad 4: 
Invertir una lista sin usar reverse() ni [::-1]

```python
from random import randint

lists_two = [randint(1, 50) for i in range(8)]

print(f"Lista original: {lists_two}")
```

```python
lists_invertida = []

for i in range(len(lists_two) - 1, -1, -1):
    lists_invertida.append(lists_two[i])

print(f"Lista invertida: {lists_invertida}")
```

### Actividad 5:
Sumar los elementos de dos listas posición por posición, utilizando solo bucles repetititivos (for,while).

```python
lists_one = [2, 4, 6, 7, 8]
lists_two = [1, 3, 5, 9, 10]

print(f"Lista 1: {lists_one}")
print(f"Lista 2: {lists_two}")
```

```python
suma_listas = []

for i in range(len(lists_one)):
    suma_listas.append(lists_one[i] + lists_two[i])

print(f"Suma de listas: {suma_listas}")
```

```python
suma_listas = []

i = 0

while i < len(lists_one):
    suma_listas.append(lists_one[i] + lists_two[i])
    i += 1

print(f"Suma de listas: {suma_listas}")
```

### Actividad 6:

Escribir un programa que almacene el abecedario en una lista, elimine de la lista las letras que ocupen posiciones múltiplos de 3, y muestre por pantalla la lista resultante.

```python
abecedario = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

resultado = [letra for i, letra in enumerate(abecedario) if (i + 1) % 3 != 0]

print(f"Lista resultante: {resultado}")
```

### Actividad 8:

Teniendo en cuenta el conjunto 1 y el conjunto 2, sin usar el operador de intersección &, calcular la intersección de dichos conjuntos.

```python
conjunto1 = {1, 2, 3, 4}
conjunto2 = {3, 4, 5, 6}

interseccion = set()

for elemento in conjunto1:
    if elemento in conjunto2:
        interseccion.add(elemento)

print(f"Intersección de los conjuntos: {interseccion}")
```

### Actividad 9:
Eliminar duplicados de una lista usando un set.

```python
lista = [1, 2, 2, 3, 4, 4, 5]
lista_sin_duplicados = list(set(lista))

print(f"Lista original: {lista}")
print(f"Lista sin duplicados: {lista_sin_duplicados}")

```

### Actividad 10:

Escribir un programa que cree 2 matrices y realice su multiplicacion.

```python
matriz_a = [
    [1, 2],
    [3, 4]
]

matriz_b = [
    [5, 6],
    [7, 8]
]

resultado = [[0 for _ in range(len(matriz_b[0]))] for _ in range(len(matriz_a))]

for i in range(len(matriz_a)):
    for j in range(len(matriz_b[0])):
        for k in range(len(matriz_b)):
            resultado[i][j] += matriz_a[i][k] * matriz_b[k][j]


print("\nMatriz resultado (A x B):")
for fila in resultado:
    print(fila)
```
